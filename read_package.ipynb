## 欄位說明

### 命名與資料解讀原則

- `i` / `o`：分別代表流入（inbound）與流出（outbound）方向，方向以資料集的監測端為基準。
- `tt`：total，表示總數或加總值；`avg`：average，表示平均值。
- `Sess_`：Session 層級的彙整或模型結果；沒有此前綴的預測欄位通常屬於封包層級。
- `None` 或空值：可能表示該協定不適用、該方向沒有流量，或該階段尚未產生模型結果。

```python
# 隨機抽取 2,000 個 Session，分別建立原始／清理後 payload 的 RoBERTa embedding，並以 UMAP 降成 3D。
import importlib.util
import subprocess
import sys

required_packages = {
    "torch": "torch",
    "transformers": "transformers",
    "umap": "umap-learn",
    "sklearn": "scikit-learn",
    "plotly": "plotly",
    "tqdm": "tqdm",
}
missing_packages = [
    package_name
    for module_name, package_name in required_packages.items()
    if importlib.util.find_spec(module_name) is None
]
if missing_packages:
    print("安裝缺少的套件：", ", ".join(missing_packages))
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "--quiet", *missing_packages]
    )
    importlib.invalidate_caches()

import numpy as np
import pandas as pd
import plotly.express as px
import torch
import torch.nn.functional as F
import umap.umap_ as umap
from tqdm import tqdm
from transformers import AutoModel, AutoTokenizer

RANDOM_STATE = 42
SAMPLE_SIZE = 3000
MODEL_NAME = "FacebookAI/roberta-base"
MAX_LENGTH = 512  # 可提高至 512 以保留更多內容，但會增加記憶體與執行時間。
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 32 if DEVICE.type == "cuda" else 8

required_columns = {"Session_ID", "payload_list", "clean_payload_list"}
missing_columns = required_columns.difference(data.columns)
if missing_columns:
    raise KeyError(f"缺少必要欄位：{sorted(missing_columns)}")
if len(data) < SAMPLE_SIZE:
    raise ValueError(f"資料只有 {len(data):,} 筆，無法抽取 {SAMPLE_SIZE:,} 筆。")

# 兩個欄位使用相同的 2,000 個 Session，讓原始與清理後 payload 可以直接比較。
sampled_payloads = (
    data.loc[:, ["Session_ID", "payload_list", "clean_payload_list"]]
    .sample(n=SAMPLE_SIZE, random_state=RANDOM_STATE, replace=False)
    .copy()
)
sampled_payloads.insert(0, "row_index", sampled_payloads.index.astype(str))
sampled_payloads.reset_index(drop=True, inplace=True)


def payload_to_text(value):
    """將 payload list、bytes 或字串統一轉成 Transformer 可讀文字。"""
    if isinstance(value, bytes):
        return value.decode("utf-8", errors="replace")
    if isinstance(value, (list, tuple, np.ndarray)):
        return "\n[PACKET]\n".join(
            payload_to_text(item) for item in value if item is not None
        )
    if value is None:
        return ""
    try:
        if pd.isna(value):
            return ""
    except (TypeError, ValueError):
        pass
    return str(value)


sampled_payloads["payload_text"] = sampled_payloads["payload_list"].map(payload_to_text)
sampled_payloads["clean_payload_text"] = sampled_payloads["clean_payload_list"].map(
    payload_to_text
)

print(f"載入模型：{MODEL_NAME}")
print(f"運算裝置：{DEVICE}；batch size：{BATCH_SIZE}")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME).to(DEVICE)
model.eval()


@torch.inference_mode()
def embed_texts(texts, description):
    """以 RoBERTa self-attention 產生上下文 token，再做 masked mean pooling。"""
    vectors = []
    for start in tqdm(range(0, len(texts), BATCH_SIZE), desc=description):
        batch_texts = texts[start : start + BATCH_SIZE]
        encoded = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_special_tokens_mask=True,
            return_tensors="pt",
        )
        special_tokens_mask = encoded.pop("special_tokens_mask").to(DEVICE)
        model_inputs = {name: tensor.to(DEVICE) for name, tensor in encoded.items()}
        outputs = model(**model_inputs)

        # 排除 padding 與特殊 token，再平均所有上下文 token 向量。
        attention_mask = model_inputs["attention_mask"]
        valid_mask = attention_mask * (1 - special_tokens_mask)
        empty_rows = valid_mask.sum(dim=1) == 0
        valid_mask[empty_rows] = attention_mask[empty_rows]
        valid_mask = valid_mask.unsqueeze(-1).type_as(outputs.last_hidden_state)
        pooled = (outputs.last_hidden_state * valid_mask).sum(dim=1)
        pooled = pooled / valid_mask.sum(dim=1).clamp(min=1e-9)
        pooled = F.normalize(pooled, p=2, dim=1)
        vectors.append(pooled.cpu().numpy())

    return np.vstack(vectors)


payload_embeddings = embed_texts(
    sampled_payloads["payload_text"].tolist(), "Embedding payload_list"
)
clean_payload_embeddings = embed_texts(
    sampled_payloads["clean_payload_text"].tolist(),
    "Embedding clean_payload_list",
)

# 共同 fit UMAP，兩個欄位的 3D 座標才位於相同空間、可直接比較。
combined_embeddings = np.vstack([payload_embeddings, clean_payload_embeddings])
umap_model = umap.UMAP(
    n_components=3,
    n_neighbors=30,
    min_dist=0.1,
    metric="cosine",
    random_state=RANDOM_STATE,
    n_jobs=1,
)
umap_coordinates = umap_model.fit_transform(combined_embeddings)

plot_data = pd.DataFrame(umap_coordinates, columns=["UMAP_1", "UMAP_2", "UMAP_3"])
plot_data["field"] = np.repeat(
    ["payload_list", "clean_payload_list"], SAMPLE_SIZE
)
plot_data["row_index"] = pd.concat(
    [sampled_payloads["row_index"], sampled_payloads["row_index"]],
    ignore_index=True,
)
plot_data["Session_ID"] = pd.concat(
    [sampled_payloads["Session_ID"], sampled_payloads["Session_ID"]],
    ignore_index=True,
)
plot_data["preview"] = pd.concat(
    [sampled_payloads["payload_text"], sampled_payloads["clean_payload_text"]],
    ignore_index=True,
).str.replace(r"\s+", " ", regex=True).str.slice(0, 180)

fig = px.scatter_3d(
    plot_data,
    x="UMAP_1",
    y="UMAP_2",
    z="UMAP_3",
    color="field",
    color_discrete_map={
        "payload_list": "#1f77b4",
        "clean_payload_list": "#ff7f0e",
    },
    hover_data={
        "row_index": True,
        "Session_ID": True,
        "preview": True,
        "UMAP_1": ":.3f",
        "UMAP_2": ":.3f",
        "UMAP_3": ":.3f",
    },
    opacity=0.65,
    title="RoBERTa Embeddings：payload_list vs. clean_payload_list（3D UMAP）",
)
fig.update_traces(marker={"size": 3})
fig.update_layout(
    width=1000,
    height=750,
    legend_title_text="欄位",
    scene={"aspectmode": "cube"},
)
print("RoBERTa embedding 與 3D UMAP 計算完成；請執行下一個 cell 顯示圖表。")
```
